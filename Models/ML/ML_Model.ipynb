```python
# Set a random seed to keep later randomized selections consistent
random_seed = 42026
np.random.seed(random_seed)

# Load in teleconnection and regional anomaly data
nh_data = xr.open_dataset("../../Data/NH_tele_cli_anoms_detrended_1950-2024.nc")
sh_data = xr.open_dataset("../../Data/SH_tele_cli_anoms_detrended_1950-2024.nc")
global_data = xr.open_dataset("../../Data/Global_tele_cli_anoms_detrended_1950-2024.nc")
total_data = xr.merge([nh_data, sh_data, global_data], compat='override')

# The original, local version of this code imported data in from a combined CSV file. To emulate this for the repository, a temporary CSV file is generated from the combined netCDF data files.
total_data.to_dataframe().to_csv("temp.csv")
total = pd.read_csv("temp.csv")

# Trim data to only necessary fields and those defined by the coorelation analysis dictionary from earlier
feature_list = ["year", "month"] + tele_dict[label]  + [f"{label}"]
total_trimmed = total[feature_list]
#total_trimmed
```

```python
# Remove Null/NA values (Note that in select regions/teleconnection patterns this may result in a significant amount of datapoints being removed)
final = total_trimmed.dropna()

# Reset indexing to match new data (Will make future work with pooling easier)
final.reset_index(drop=True, inplace=True)
#final
```

```python
#Import data values from trimmed dataset
df_temp = final

#Create empty array of years
yr_range = []

# Fill the year range array with all of the year values from the trimmed input dataset
for i in range(0,len(final["year"])):
    if i == 0:
        yr_range.append(final["year"][i])
    else:
        year_flag = 0
        for j in yr_range:
            if final["year"][i] == j:
                year_flag = 1
        if year_flag == 0:
            yr_range.append(final["year"][i])

#TROUBLESHOOTING: Print final year range array (Can be used to verify the function scraped the years properly
#print(yr_range)

#Count the number of years in the year range array
yr_count = len(yr_range)

# Create training and testing dataset pool sizes based on the number of years (By default a 80/20 split is used since validation dataset pool is handled automatically by YDF)
train_count = round((yr_count * 0.8))
test_count = round(yr_count * 0.2)

#TROUBLESHOOTING: Print the pool sizes as well as the total to validate the pool counts/rounding
print(f"train: {train_count}, test: {test_count}, total: {yr_count}")
```

```python
#Create a set of loops that randomly selects entries via RNG (random number generation) of years
test_nums = []
train_nums = []
c = 1

while c <= train_count:
    temp_num = np.random.choice(yr_range)
    train_nums.append(temp_num)
    yr_range.remove(temp_num)
    c += 1

c = 1

while c <= test_count:
    temp_num = np.random.choice(yr_range)
    test_nums.append(temp_num)
    yr_range.remove(temp_num)
    c += 1

# TROUBLESHOOTING: Print the randomly generated list of years for each pool
#print(test_nums)
#print(train_nums)
```

```python
# Using the newly generated pool of years for training and testing data, import data from trimmed dataset for each of those pools and combine them into training and testing datasets
for i in train_nums:
    if i == train_nums[0]:
        df_train = final[final["year"] == i]
    else:
        df_train = pd.concat([df_train, final[final["year"] == i]])
df_train.sort_values(by=["year"], inplace=True)

for i in test_nums:
    if i == test_nums[0]:
        df_test = final[final["year"] == i]
    else:
        df_test = pd.concat([df_test, final[final["year"] == i]])
df_test.sort_values(by=["year"], inplace=True)

df_train
```

```python
#Filter out the now un-necessary year variable (Note that since we're utilizing detrended data, year should not be significant. If using normal data, consider including it)
subset1_list = feature_list[1:]
subset1 = df_train[subset1_list]

subset2_list = feature_list[1:]
subset2 = df_test[subset2_list]
```

```python
# Train the random forest model using the teleconnection indices, month, and variable defined in "label"

# Enable hyperparameter tuning using the default settings (See YDF documentation for more details)
tuner = ydf.RandomSearchTuner(num_trials=50, automatic_search_space=True)

# Created the Random Forest Learner model using our label variable, newly defined tuner, and random seed. (Regression is selected for comparison with MLR model, can be set to classification for a more operationally optimized situation)
model = ydf.RandomForestLearner(label=label, task=ydf.Task.REGRESSION, tuner=tuner, random_seed=random_seed, working_dir="./modelTemp/").train(subset1)
```

```python
#Print RF model information
model.describe()
```

```python
# Print RF model evaluation information and figures

evaluation = model.evaluate(subset2)

# Query RMSE information
print(f"Test RMSE: {evaluation.rmse}")

# Print full evaluation report with figures
print("Full evaluation report:")
evaluation
```
