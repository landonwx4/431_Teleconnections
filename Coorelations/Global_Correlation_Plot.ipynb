### Computing and Plotting Correlations

Running this widget creates and saves 13 plots (all months + individual months) of Pearson correlations on a global map for each region.

```python
warnings.filterwarnings("ignore")

month_names = {
    1: 'January', 2: 'February', 3: 'March', 4: 'April',
    5: 'May', 6: 'June', 7: 'July', 8: 'August',
    9: 'September', 10: 'October', 11: 'November', 12: 'December'
}

dataset_options = {
    'Northern Hemisphere': nh_ds,
    'Southern Hemisphere': sh_ds,
    'Global': global_ds
}

index_map = {
    'Northern Hemisphere': ['TNA', 'WPO', 'WP', 'SCA', 'EAWR', 'EA', 'ADI', 'AMO',
                            'AO', 'EPNP', 'EPO', 'NAO', 'NOI', 'PDO', 'PNA', 'POL'],
    'Southern Hemisphere': ['SAOD', 'TSA', 'SPOD', 'TPI', 'SOI', 'AAO'],
    'Global': ['IPO', 'IOD', 'EMI', 'QBO', 'MEI', 'AAM', 'ENSO_34']
}

regions = [
    [-98, -81, 37, 49, 'Midwest'],
    [-105, -93, 26, 41, 'Southern_Plains'],
    [-81, -67, 37, 48, 'Northeast'],
    [-9, 11, 36, 55, 'Western_Europe'],
    [11, 32, 36, 55, 'Eastern_Europe'],
    [32, 55, 38, 57, 'Black_Sea'],
    [65, 90, 5, 33, 'India'],
    [105, 130, 20, 45, 'East_Asia'],
    [111, 135, -36, -12, 'Western_Australia'],
    [135, 155, -40, -14, 'Eastern_Australia'],
    [-60, -35, -15, 0, 'Northern_Brazil'],
    [-60, -35, -30, -15, 'Southern_Brazil'],
    [-69, -55, -22, -42, 'Argentina']
]

month_options = [('All Months', None)] + [(f"{month_names[m]}", m) for m in range(1, 13)]

def get_variable_name(base_var, region_name):
    return f"{base_var}_StdAnom_{region_name.replace(' ', '_')}_Detrended"

def compute_and_plot(dataset_name, index_name, month, variable_type):

    # Selecting the data
    ds = dataset_options[dataset_name]
    base_var = 'T2M' if variable_type == 'Std Temp Anomaly' else 'TP'
    title_month = "All Months" if month is None else month_names[month]
    data_full = ds if month is None else ds.where(ds.time.dt.month == month, drop=True)

    correlations = []

    region_names = [r[4] for r in regions]

    # Comuting correlations in each region
    for region_name in region_names:

        var_name = get_variable_name(base_var, region_name)

        data_da = data_full[var_name]

        d = data_da.values
        i = ds[index_name].values

        dt = data_da.time.values
        it = ds[index_name].time.values

        # Matching overlapping timestamps between teleconnection and anomaly data
        m1 = np.isin(dt, it)
        m2 = np.isin(it, dt[m1])

        x = d[m1]
        y = i[m2]

        v = ~(np.isnan(x) | np.isnan(y))

        r = pearsonr(x[v], y[v])[0] if v.sum() >= 5 and m1.sum() >= 5 else 0.0

        correlations.append(abs(r))

    max_abs = max(correlations) if correlations else 0.3
    norm = colors.Normalize(vmin=-max_abs, vmax=max_abs)

    # Plotting
    fig, ax = plt.subplots(figsize=(10, 8), dpi=400, subplot_kw={'projection': ccrs.PlateCarree()})

    title_var = "Temperature" if variable_type == 'Std Temp Anomaly' else "Precipitation"
    ax.set_title(f'{index_name} Correlation with {title_var} ({title_month})', fontsize=14, fontweight='bold')

    cmap = (plt.colormaps['seismic'] if variable_type == 'Std Temp Anomaly' else plt.colormaps['BrBG'])

    for lon_min, lon_max, lat_min, lat_max, region_name in regions:

        var_name = get_variable_name(base_var, region_name)

        data_da = data_full[var_name]

        d = data_da.values
        i = ds[index_name].values

        dt = data_da.time.values
        it = ds[index_name].time.values

        m1 = np.isin(dt, it)
        m2 = np.isin(it, dt[m1])

        x = d[m1]
        y = i[m2]

        v = ~(np.isnan(x) | np.isnan(y))

        region_corr = (pearsonr(x[v], y[v])[0] if v.sum() >= 5 and m1.sum() >= 5 else 0.0)

        color = cmap(norm(region_corr))

        # This outlines the regions on the map
        rect = Rectangle(xy=(lon_min, lat_min), width=lon_max-lon_min, height=lat_max-lat_min, facecolor=color,
                         edgecolor='black', linewidth=1.8, alpha=1, transform=ccrs.PlateCarree(), zorder=2)

        ax.add_patch(rect)

        cx = (lon_min + lon_max) / 2
        cy = (lat_min + lat_max) / 2

        # Printing the correlation value in the middle of each box
        ax.text(cx, cy, f"{region_corr:.2f}", transform=ccrs.PlateCarree(), ha='center', va='center', fontsize=8, fontweight='bold',
                color='white', zorder=3, path_effects=[patheffects.withStroke(linewidth=1.5, foreground='black')])

    ax.add_feature(cfeature.COASTLINE, lw=0.7, color='dimgray', zorder=2)
    ax.add_feature(cfeature.LAND, facecolor='#f0f0e8', zorder=0)
    ax.add_feature(cfeature.OCEAN, facecolor='white', zorder=0)
    gl = ax.gridlines(draw_labels=True, linestyle="--", linewidth=0.4, color="gray", alpha=0.7, zorder=1)
    gl.top_labels = False
    gl.right_labels = False
    ax.set_extent([-180, 180, -90, 90], crs=ccrs.PlateCarree())

    sm = cm.ScalarMappable(cmap=cmap, norm=norm)
    cbar = fig.colorbar(sm, ax=ax, orientation='vertical', pad=0.01, shrink=0.506, aspect=20)
    cbar.set_label(f'Correlation with {title_var}', fontsize=11)

    plt.tight_layout()
    display(fig)

    var_short = ("Temp" if variable_type == 'Std Temp Anomaly' else "Precip")

    filename = (f"{index_name}_{title_month.replace(' ', '_')}_{var_short}_Correlation.png")
    fig.savefig(filename, dpi=400, bbox_inches='tight')
    plt.close(fig)

dataset_widget = widgets.Dropdown(
    options=list(dataset_options.keys()),
    value='Global',
    description='Dataset:'
)

index_widget = widgets.Dropdown(
    options=index_map['Global'],
    description='Teleconnection:'
)

# This conveniently updates the teleconnection list based upon the hemisphere you select
def update_index_options(change):
    index_widget.options = index_map[change['new']]

dataset_widget.observe(update_index_options, names='value')

variable_widget = widgets.Dropdown(
    options=['Std Temp Anomaly', 'Std Precip Anomaly'],
    value='Std Temp Anomaly',
    description='Variable:'
)

button = widgets.Button(
    description="Plot Correlation Maps",
    button_style='success',
    layout=widgets.Layout(width='200px')
)

output = widgets.Output()

def on_button_clicked(b):

    with output:
        clear_output(wait=True)

        # Looping through the 12 individual months + all months averaged
        for month_label, month in month_options:
            print(f"\n{month_label}\n")
            try:

                # Computing correlations and plotting
                compute_and_plot(
                    dataset_widget.value,
                    index_widget.value,
                    month,
                    variable_widget.value
                )

            # Some months in the teleconnection dataset are all NaNs, so we must ignore
            except ValueError as e:
                if "Found array with 0 sample(s)" in str(e):
                    print(f"No valid data available for {month_label}.")
                else:
                    raise

button.on_click(on_button_clicked)

# Calling the widget
display(
    widgets.VBox([
        dataset_widget,
        index_widget,
        variable_widget,
        button,
        output
    ])
)
```

```python
regions = [
    [-98, -81, 37, 49, -0.447, 'EPNP'], # Midwest
    [-105, -93, 26, 41, -0.278, 'EPNP'], # Southern Plains
    [-81, -67, 37, 48, -0.364, 'EPNP'], # Northeast
    [-9, 11, 36, 55, 0.366, 'EA'], # Western Europe
    [11, 32, 36, 55,  0.408, 'EA'], # Eastern Europe
    [32, 55, 38, 57, -0.467, 'EAWR'], # Black Sea Region
    [65, 90, 5, 33, 0.212, 'PDO'], # India
    [105, 130, 20, 45, 0.236, 'EA'], # East Asia
    [111, 135, -36, -12, 0.216, 'ENSO3.4'], # Western Australia
    [135, 155, -40, -14, -0.209, 'AAO'], # Eastern Australia
    [-60, -35, -15, 0, 0.274, 'ENSO3.4'], # Northern Brazil
    [-60, -35, -30, -15, 0.269, 'TPI'], # Southern Brazil
    [-69, -55, -22, -42, 0.140, 'TPI'] # Argentina
]

values = [item[4] for item in regions]
norm = Normalize(vmin=-0.45, vmax=0.45)
cmap = plt.colormaps['seismic']

fig, ax = plt.subplots(figsize=(10, 8), dpi = 400, subplot_kw = {'projection': ccrs.PlateCarree()})
ax.set_title('Strongest Index Correlation for All Months (Temperature)', fontsize=12, fontweight='bold')

for lon_min, lon_max, lat_min, lat_max, val, index in regions:
    color = cmap(norm(val))

    rect = Rectangle(xy=(lon_min, lat_min), width=lon_max-lon_min, height=lat_max-lat_min,
                             facecolor=color, edgecolor='black', linewidth=1.8, alpha=1,
                             transform=ccrs.PlateCarree(), zorder=2)
    ax.add_patch(rect)

    center_lon = (lon_min + lon_max) / 2
    center_lat = (lat_min + lat_max) / 2

    ax.text(
        center_lon, center_lat,
        f"{index}",
        transform=ccrs.PlateCarree(),
        ha='center',
        va='center',
        fontsize=7,
        fontweight='bold',
        color='white',
        zorder=3,

        path_effects=[
            patheffects.withStroke(
                linewidth=1,
                foreground='black'
            )
        ]
    )

sm = cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = fig.colorbar(sm, ax=ax, orientation='vertical', pad=0.01, shrink=0.528, aspect=20)
cbar.set_label('Correlation with Temperature', fontsize=10)

ax.add_feature(cfeature.COASTLINE, lw=0.7, color='dimgray', zorder=2)
ax.add_feature(cfeature.LAND, facecolor='#f0f0e8', zorder=0)
ax.add_feature(cfeature.OCEAN, facecolor='white', zorder=0)

gl = ax.gridlines(draw_labels=True, linestyle="--", linewidth=0.4, color="gray", alpha=0.7, zorder=1)
gl.top_labels = False
gl.right_labels = False
ax.set_extent([-180, 180, -90, 90], crs=ccrs.PlateCarree())
# plt.savefig(f"All_Months_Temperature_Global_MaxCorr.png", dpi=400, bbox_inches='tight')
```
